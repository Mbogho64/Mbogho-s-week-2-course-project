```python
# Function to convert temperature from Celsius to Fahrenheit
def celsius_to_fahrenheit(celsius):
  fahrenheit = (celsius * 9/5) + 32
  return fahrenheit
```

```python
# Ask the user to enter a temperature in Celsius
temperature_c = float(input("Enter temperature in °C: "))

# Convert Celsius to Fahrenheit
temperature_f = celsius_to_fahrenheit(temperature_c)

# Display the result
print(f"{temperature_c}°C is equal to {temperature_f:.2f}°F")
```

```python
# Set the temperature threshold
threshold = 35

# Check whether the temperature is above or below the threshold
if temperature_c > threshold:
    print(f"The temperature is above the threshold of {threshold}°C.")
elif temperature_c < threshold:
    print(f"The temperature is below the threshold of {threshold}°C.")
else:
    print(f"The temperature is exactly at the threshold of {threshold}°C.")
```

```python
# List of temperatures to convert
temperatures = [5, 15, 25, 35, 45, 55]

# Convert and display each temperature
for temp in temperatures:
    fahrenheit = celsius_to_fahrenheit(temp)
    print(f"{temp}°C = {fahrenheit:.2f}°F")
```

```python
# Set the temperature threshold
threshold = 35

# Convert several temperatures and compare them with the threshold
temperatures = [5, 15, 25, 35, 45, 55]

for temp in temperatures:
    fahrenheit = celsius_to_fahrenheit(temp)

    if temp > threshold:
        status = "above the threshold"
    elif temp < threshold:
        status = "below the threshold"
    else:
        status = "at the threshold"

    print(f"{temp}°C = {fahrenheit:.2f}°F ({status})")
```
